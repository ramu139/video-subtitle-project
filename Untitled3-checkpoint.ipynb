```python
import os   # handling the files
import pickle # storing numpy features
import numpy as np
from tqdm.notebook import tqdm # how much data is process till now
from keras.applications.vgg16 import VGG16 , preprocess_input # extract features from image data.
from keras.preprocessing.image import load_img , img_to_array
from tensorflow.keras.preprocessing.text import Tokenizer
from keras.preprocessing.sequence import pad_sequences
from keras.models import Model
from keras.utils import to_categorical
from keras.layers import Input, Dense, LSTM, Embedding, Dropout, add
import pandas as pd
from keras.callbacks import ModelCheckpoint
```

```python
features = np.load("/content/drive/MyDrive/model/features.npy")
captions = np.load("/content/drive/MyDrive/model/caption.npy")
features = np.reshape(features, (features.shape[0], features.shape[1] * features.shape[2]))
print(features.shape)
```

```python
tokenizer = Tokenizer()
tokenizer.fit_on_texts(captions)
vocab_size = len(tokenizer.word_index) + 1
max_length = max(len(caption.split()) for caption in captions)

def prepare_data(image_features, caption_features):
    x1, x2, y = [], [], []
    for m in range(len(image_features)):
        caption = caption_features[m]
        caption = caption.split()
        seq = tokenizer.texts_to_sequences([caption])[0]
        length = len(seq)
        for i in range(1, length):
            x2_seq, y_seq = seq[:i] , seq[i]
            x2_seq = pad_sequences([x2_seq], maxlen = max_length, padding='post')[0]
            y_seq = to_categorical([y_seq], num_classes = vocab_size)[0]
            x1.append(image_features[m])
            x2.append(x2_seq)
            y.append(y_seq)
    return np.array(x1), np.array(x2), np.array(y)
```

```python
train_x1, train_x2, train_y = prepare_data(features[0:801], captions[0:801])
print(str(train_x1.shape)+" "+str(train_x2.shape)+" "+str(train_y.shape))
validate_x1, validate_x2, validate_y = prepare_data(features[801:len(features)], captions[801:len(features)])
print(str(validate_x1.shape)+" "+str(validate_x2.shape)+" "+str(validate_y.shape))
```

```python
from keras.optimizers import Adam
from keras.layers import Bidirectional
from keras import regularizers
inputs1 = Input(shape=(4096,))
fe1 = Dropout(0.4)(inputs1)
fe2 = Dense(256, activation='relu')(fe1)
# sequence feature layers
inputs2 = Input(shape=(max_length,))
se1 = Embedding(vocab_size, 256, mask_zero=True)(inputs2)
se2 = Dropout(0.4)(se1)
se3 = LSTM(256)(se2)

# decoder model
decoder1 = add([fe2, se3])
decoder2 = Dense(256, activation='relu')(decoder1)
outputs = Dense(vocab_size, activation='softmax')(decoder2)

cnn_lstm_model = Model(inputs=[inputs1, inputs2], outputs=outputs)
optimizer = Adam(0.0001)
cnn_lstm_model.compile(loss='categorical_crossentropy', optimizer=optimizer, metrics = ['accuracy'])
if os.path.exists("/content/drive/MyDrive/model/cnn_weights.keras") == False:
    model_check_point = ModelCheckpoint(filepath='/content/drive/MyDrive/model/cnn_weights.keras', verbose = 1, save_best_only = True)
    hist = cnn_lstm_model.fit([train_x1, train_x2], train_y, verbose = 1, batch_size = 32, epochs = 120, callbacks=[model_check_point], validation_data=([validate_x1, validate_x2], validate_y))
    f = open('/content/drive/MyDrive/model/cnn_history.pckl', 'wb')
    pickle.dump(hist.history, f)
    f.close()
else:
    cnn_lstm_model.load_weights("/content/drive/MyDrive/model/cnn_weights.keras")
```

```python
score = cnn_lstm_model.evaluate([validate_x1, validate_x2], validate_y, verbose=1)
print('Test loss:', score)
print('Test accuracy:', score)
```
